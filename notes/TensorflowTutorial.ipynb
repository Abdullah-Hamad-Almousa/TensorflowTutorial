# Tensorflow 2 MNIST Demo

```python
import tensorflow as tf
```

```python
mnist = tf.keras.datasets.mnist
(x_train, y_train),(x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0
```

Build model, choose an optimizer and the loss function

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10)
])
```

Looking at what the model is doing in terms of the outputs/vectors

```python
pred = model(x_train[:1]).numpy()
pred
```

Probability of each class instead of the vectors

```python
tf.nn.softmax(pred).numpy
```
